```python
#Clean the Data
```

```python
# Start with a copy so we don't lose the original
df_clean=df.copy()
```

```python
# 1. Remove cancellations (Invoice starting with 'C')
df_clean=df_clean[~df_clean['Invoice'].astype(str).str.startswith('C')]
```

```python
# 2. Remove missing Customer ID
df_clean = df_clean.dropna(subset=['Customer ID'])
```

```python
# 3. Remove negative or zero quantity
df_clean = df_clean[df_clean['Quantity'] > 0]
```

```python
# 4. Remove zero or negative price
df_clean = df_clean[df_clean['Price'] > 0]
```

```python
# 5. Remove non-product StockCodes
non_product_codes = ['POST', 'DOT', 'M', 'D', 'C2', 'BANK CHARGES', 'B', 'PADS', 'CRUK', 'TEST']
df_clean = df_clean[~df_clean['StockCode'].astype(str).isin(non_product_codes)]
```

```python
# 6. Remove duplicates
df_clean = df_clean.drop_duplicates()
```

```python
# 7. Convert Customer ID to integer (no more decimals)
df_clean['Customer ID'] = df_clean['Customer ID'].astype(int)
```

```python
# 8. Add TotalPrice column
df_clean['TotalPrice'] = round(df_clean['Quantity'] * df_clean['Price'],2)
```

```python
print(f"Rows before cleaning: {len(df)}")
print(f"Rows after cleaning:  {len(df_clean)}")
print(f"Removed: {len(df) - len(df_clean)} rows ({(len(df)-len(df_clean))/len(df)*100:.2f}%)")
```
